```python
import numpy as np
import matplotlib.pyplot as plt
```

Lue tiedostot centers.txt, forwards.txt ja guards.txt käyttäen numpy genfromtxt()-funktiota.

```python
centers = np.genfromtxt('centers.txt', delimiter=';')
forwards = np.genfromtxt('forwards.txt', delimiter=';')
guards = np.genfromtxt('guards.txt', delimiter=';')

print(centers.shape[0], forwards.shape[0], guards.shape[0])
```

Luo taulukko categories, jossa on pelaajakategoriat seuraavasti

```python
categories = ['centers', 'forwards', 'guards']
print(categories)
```

Muodosta eri kategorioiden pelaajatiedoista kolmiulotteinen taulukko players pinoamalla (stack).

```python
players = np.stack((centers, forwards, guards))
print(players.shape)
```

Tulosta pelaajien maksimipituus ja -paino käyttäen players-taulukkoa.

```python
print(players[:, :, 0].max())
print(players[:, :, 1].max())
```

Tulosta kunkin pelaajakategorian pelaajien keskipituus käyttäen taulukkoa players.

```python
for i in range(len(categories)):
    print(categories[i], players[i, :, 0].mean())
```

Tee pelaajien pituuksista ja painoista scatterplot, jossa eri kategorioiden pelaajat on merkitty eri väreillä. Käytä taulukkoa players.

```python
for i in range(len(categories)):
    plt.scatter(players[i, :, 0], players[i, :, 1], label=categories[i])
plt.legend()
```

Muodosta players-taulukosta eri kategorioiden pelaajien pituuksien boxplot. Missä kategoriassa on suurin vaihteluväli?

```python
plt.boxplot(players[:, :, 0].T)
plt.show()
# Suurin vaihteluväli on forwards-kategoriassa (kategoria 2: 194-218 cm)
```
